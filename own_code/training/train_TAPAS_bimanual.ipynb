```python
from datetime import date
from pathlib import Path
import os
import random

from tapas_gmm.dataset.scene import SceneDataset
from tapas_gmm.dataset.bc import BCDataset, BCDataConfig
from torch.utils.data import DataLoader
from tapas_gmm.utils.observation import collate
from tapas_gmm.dataset.demos import Demos
from tapas_gmm.utils.random import configure_seeds

from tapas_gmm.policy.models.tpgmm import (
    AutoTPGMM,
    AutoTPGMMConfig,
    TPGMMConfig,
    FrameSelectionConfig,
    DemoSegmentationConfig,
    FittingStage,
)
```

```python
loaded_dataset = SceneDataset(
    data_root=Path(data_root)
)
```

```python
bc_config = BCDataConfig(
    fragment_length = -1,
    cameras = tuple(),
)

bc_dataset = BCDataset(
    scene_dataset=loaded_dataset,
    config=bc_config,
)
```

```python
tpgmm_config = TPGMMConfig(
    n_components = 10,
    add_time_component = True,
    position_only = False,
    add_action_component = False,
    add_gripper_action=True,
    antipodal_rotations=True,
    fixed_first_component_n_steps=1,

    reg_diag = 2e-4,
    reg_em_finish_diag = 2e-4,
    reg_init_diag = 5e-4,

    reg_em_finish_diag_gripper = 2e-2,
    reg_diag_gripper = 2e-2,
    reg_em_trans = 1e-6,
)

segmentation_config = DemoSegmentationConfig(
    distance_based = False,
    velocity_based = phase_alignment == "velocity",
    gripper_based = phase_alignment == "gripper",
    components_prop_to_len = True,
    min_n_components = 3,
)

frame_selection_config = FrameSelectionConfig(
    fix_frames = True,
    rel_score_threshold = 0.75,
    fitting_actions = [FittingStage.INIT],
    drop_redundant_frames = True,
)

auto_config = AutoTPGMMConfig(
    tpgmm=tpgmm_config,
    frame_selection=frame_selection_config,
    demos_segmentation=segmentation_config,
)
```

```python
import numpy as np
from tapas_gmm.policy.models.tpgmm import get_indeces_of_duplicate_rows


def select_frames_temporal(self, demos, fix_frames=True, init_strategy=None,
                           fitting_actions=None, rel_score_threshold=0.75):
    """Frames pro Zeitabschnitt nach Position und gesamter Pose vergleichen."""
    if not demos.is_bimanual or not self.config.frame_selection.use_precision:
        raise ValueError("Diese Frame-Auswahl benötigt bimanuale Demos und use_precision=True.")
    candidates, _ = self._fit_frames(demos, fix_frames, init_strategy, fitting_actions)
    start = int(self.config.tpgmm.add_time_component)

    def log_precision(covariance):
        sign, logdet = np.linalg.slogdet(covariance)
        if sign <= 0 or not np.isfinite(logdet):
            raise ValueError("Ungültige Kovarianz bei der Frame-Auswahl.")
        return -logdet

    full_logs, xyz_logs = [], []
    for frame in candidates:
        full_arm, xyz_arm = [], []
        for arm in frame:
            full_component, xyz_component = [], []
            for gaussian in arm.model.gaussians:
                covariance = gaussian.sigma
                full_component.append(log_precision(covariance))
                xyz_component.append(log_precision(covariance[start:start + 3, start:start + 3]))
            full_arm.append(full_component)
            xyz_arm.append(xyz_component)
        full_logs.append(full_arm)
        xyz_logs.append(xyz_arm)

    full_logs, xyz_logs = np.asarray(full_logs), np.asarray(xyz_logs)
    full = np.exp(full_logs - full_logs.max(axis=0, keepdims=True))
    xyz = np.exp(xyz_logs - xyz_logs.max(axis=0, keepdims=True))
    relative = np.maximum(full, xyz).max(axis=2).T

    metric_rows = np.concatenate((full, xyz), axis=2).reshape(len(candidates), -1)
    duplicates = get_indeces_of_duplicate_rows(metric_rows)
    redundant = [int(i) for group in duplicates for i in group[1:]]
    selected = relative > rel_score_threshold
    if self.config.frame_selection.drop_redundant_frames:
        selected[:, redundant] = False
    indices = tuple(tuple(int(i) for i in np.flatnonzero(arm)) for arm in selected)
    if not all(indices):
        raise ValueError("Für mindestens einen Arm wurde kein Frame ausgewählt.")

    self._frame_selection_rule = "temporal_xyz_pose_precision_deduplicated_v1"
    if not hasattr(self, "frame_selection_audit"):
        self.frame_selection_audit = []
    self.frame_selection_audit.append(dict(
        frame_names=list(demos.frame_names), selected=[list(i) for i in indices],
        redundant=redundant, relative_scores=relative.tolist(), threshold=rel_score_threshold))
    return -relative, relative, indices


model = AutoTPGMM(auto_config)
```

```python
original_select_frames = AutoTPGMM._select_frames
AutoTPGMM._select_frames = select_frames_temporal
try:
    lik, avg_loglik = model.fit_trajectories(
        demos = demos,
        fix_frames=data_root.name == "bimanual_dual_push_buttons",
        global_frames=False,
        fitting_actions = [ 
            FittingStage.INIT,
            FittingStage.EM_HMM,
        ],
    )
finally:
    AutoTPGMM._select_frames = original_select_frames
```

```python
def train_tapas_variant(plan_path, task, demos, seed):
    import json, hashlib, pickle, time, traceback
    from copy import deepcopy
    import numpy as np
    plan = json.loads(Path(plan_path).read_text())
    project = Path(plan["project"])
    for name, checksum in plan["source_sha256"].items():
        assert hashlib.sha256((project / name).read_bytes()).hexdigest() == checksum, name
    cells = json.loads((project / "own_code/notebooks/train_TAPAS_bimanual.ipynb").read_text())["cells"][:11]
    assert hashlib.sha256(json.dumps([c["source"] for c in cells], sort_keys=True).encode()).hexdigest() == plan["training_cells_sha256"]
    output = Path(plan["model_dir"]) / task / "tapas" / f"{demos}demos" / f"seed{seed}"
    state_file = Path(plan["directory"]) / "states" / f"{task}_{demos}_{seed}.json"
    if state_file.exists():
        state = json.loads(state_file.read_text())
        assert state["status"] == "completed", "Inspect the previous attempt before retrying"
        assert hashlib.sha256((output / "latest.pkl").read_bytes()).hexdigest() == state["checkpoint_sha256"]
        return state
    output.parent.mkdir(parents=True, exist_ok=True)
    state_file.parent.mkdir(exist_ok=True)
    scope = {}
    os.environ.update(TRAIN_TASK=plan["tasks"][task], TRAIN_DEMOS=str(demos), TRAIN_SEED=str(seed),
                      TRAIN_OUTPUT=str(output), TAPAS_PHASE_ALIGNMENT="gripper")
    state = dict(task=task, model="tapas", demonstrations=demos, demos=demos, seed=seed,
                 status="training", started=time.time(), pid=os.getpid(), cohort=plan["cohort"],
                 source_sha256=plan["source_sha256"], training_cells_sha256=plan["training_cells_sha256"])
    def save():
        temporary = state_file.with_suffix(".tmp")
        temporary.write_text(json.dumps(state, indent=2))
        temporary.replace(state_file)
    save()
    try:
        for index, cell in enumerate(cells):
            if task == "push_button" and 2 <= index <= 6:
                if index != 6:
                    continue
                source = Path(plan["directory"]) / "demo_sources" / f"{demos}demos/latest.pkl"
                assert hashlib.sha256(source.read_bytes()).hexdigest() == plan["push_demo_checkpoints"][str(demos)]
                saved = pickle.load(source.open("rb"))._demos
                assert saved.n_trajs == demos
                restored = deepcopy(saved)
                order = list(range(demos))
                random.shuffle(order)
                for name, value in vars(restored).copy().items():
                    if isinstance(value, tuple) and len(value) == demos and name != "frame_names":
                        setattr(restored, name, tuple(value[i] for i in order))
                    elif name.startswith("stacked_"):
                        setattr(restored, name, value[order])
                restored.subsample_to_common_length()
                scope["demos"] = restored
                state["demo_source_checkpoint"] = str(source)
                state["demo_source_checkpoint_sha256"] = plan["push_demo_checkpoints"][str(demos)]
            else:
                exec(compile("".join(cell["source"]), f"<train-notebook-cell-{index}>", "exec"), scope)
            state["cell"] = index
            state["updated"] = time.time()
            save()
        for name, checksum in plan["source_sha256"].items():
            assert hashlib.sha256((project / name).read_bytes()).hexdigest() == checksum, name
        checkpoint = output / "latest.pkl"
        state.update(status="completed", training_seconds=time.time() - state["started"],
                     checkpoint_sha256=hashlib.sha256(checkpoint.read_bytes()).hexdigest(),
                     checkpoint=str(checkpoint), segments=len(scope["model"].segment_gmms),
                     online=False, data_rule="Original nested demonstrations; no generated/reconstructed model predictions")
        (output / "run.json").write_text(json.dumps(state, indent=2))
    except BaseException:
        state.update(status="failed", error=traceback.format_exc())
        raise
    finally:
        save()
    return state
```

```python
def publish_tapas_variant(plan, state, github_token):
    import json, hashlib, zipfile, urllib.request, urllib.error
    from datetime import datetime, timezone
    directory = Path(plan["directory"])
    for p, checksum in plan["source_sha256"].items():
        assert hashlib.sha256((Path(plan["project"]) / p).read_bytes()).hexdigest() == checksum, p
    model = Path(state["checkpoint"])
    assert hashlib.sha256(model.read_bytes()).hexdigest() == state["checkpoint_sha256"]
    name = f'{state["task"]}_tapas_{state["demos"]}demos_seed{state["seed"]}.zip'
    receipts = directory / "online"
    receipts.mkdir(exist_ok=True)
    receipt = receipts / (name + ".json")
    api = "https://api.github.com/repos/Nils19024/Study-Project"
    headers = {"Authorization": "Bearer " + github_token, "User-Agent": "Study-Project-backup",
               "Accept": "application/vnd.github+json"}
    def request(url, data=None, content_type="application/json"):
        head = dict(headers, **{"Content-Type": content_type})
        payload = json.dumps(data).encode() if isinstance(data, dict) else data
        with urllib.request.urlopen(urllib.request.Request(url, data=payload, headers=head), timeout=300) as response:
            return json.load(response)
    try:
        release = request(api + "/releases/tags/" + plan["release_tag"])
    except urllib.error.HTTPError as error:
        if error.code != 404:
            raise
        release = request(api + "/releases", dict(tag_name=plan["release_tag"],
            name="TAPAS: corrected 25 and 100 demo models",
            body="Separate corrected-training cohort. Original demonstrations, three training seeds per configuration. Original older models remain in checkpoints-2026-09-25. Each ZIP contains the original checkpoint, run.json and training-source provenance. Only uploaded ZIPs are secured; the cohort can be incomplete while fitting continues.",
            draft=False, prerelease=False))
    archive_dir = directory / "archives"
    archive_dir.mkdir(exist_ok=True)
    archive = archive_dir / name
    if not archive.exists():
        files = {f'training_runs/{state["task"]}/tapas/{state["demos"]}demos/seed{state["seed"]}/latest.pkl': model,
                 f'training_runs/{state["task"]}/tapas/{state["demos"]}demos/seed{state["seed"]}/run.json': model.with_name("run.json")}
        files.update({"provenance/" + p: Path(plan["project"]) / p for p in plan["source_sha256"]})
        files["provenance/train_TAPAS_bimanual.ipynb"] = Path(plan["project"]) / "own_code/notebooks/train_TAPAS_bimanual.ipynb"
        contents = {p: dict(bytes=f.stat().st_size, sha256=hashlib.sha256(f.read_bytes()).hexdigest()) for p, f in files.items()}
        temporary = archive.with_suffix(".part")
        with zipfile.ZipFile(temporary, "x", zipfile.ZIP_DEFLATED, compresslevel=1, allowZip64=True) as z:
            for p, f in files.items():
                z.write(f, p)
            z.writestr("checkpoint_manifest.json", json.dumps(dict(cohort=plan["cohort"], files=contents), indent=2))
        temporary.replace(archive)
    with zipfile.ZipFile(archive) as z:
        contents = json.loads(z.read("checkpoint_manifest.json"))["files"]
        for p, meta in contents.items():
            content = z.read(p)
            assert len(content) == meta["bytes"] and hashlib.sha256(content).hexdigest() == meta["sha256"]
    checksum = hashlib.sha256(archive.read_bytes()).hexdigest()
    assets = request(api + f'/releases/{release["id"]}/assets?per_page=100')
    asset = next((x for x in assets if x["name"] == name), None)
    if asset is None:
        asset = request(f'https://uploads.github.com/repos/Nils19024/Study-Project/releases/{release["id"]}/assets?name={name}',
                        archive.read_bytes(), "application/zip")
    asset = request(api + "/releases/assets/" + str(asset["id"]))
    assert asset["state"] == "uploaded" and asset["size"] == archive.stat().st_size
    assert asset["digest"] == "sha256:" + checksum
    result = dict(asset_id=asset["id"], url=asset["browser_download_url"], sha256=checksum,
                  checkpoint_sha256=state["checkpoint_sha256"], verified_utc=datetime.now(timezone.utc).isoformat(),
                  bytes=archive.stat().st_size, release=release["html_url"])
    receipt.write_text(json.dumps(result, indent=2))
    return result


def run_tapas_training_queue(plan_path, github_token):
    import json, time, hashlib, subprocess, sys, signal, socket, fcntl
    from datetime import datetime
    plan_path = Path(plan_path)
    plan = json.loads(plan_path.read_text())
    directory = Path(plan["directory"])
    lock = (directory / "training.lock").open("a+")
    fcntl.flock(lock, fcntl.LOCK_EX | fcntl.LOCK_NB)
    lock.write(str(os.getpid()))
    lock.flush()
    notebook = Path(plan["project"]) / "own_code/notebooks/train_TAPAS_bimanual.ipynb"
    code = "import os,random,json,sys\nfrom pathlib import Path\n"
    code += f"sys.path[:0] = {[plan['project'] + '/' + x for x in ('TAPAS','riepybdlib')]!r}\n"
    code += f"nb=json.load(open({str(notebook)!r}))\n"
    code += "exec(''.join(next(c['source'] for c in nb['cells'] if c.get('id')=='tapas-benchmark-training')))\n"
    code += "train_tapas_variant(*json.loads(sys.argv[1]))\n"
    env = dict(os.environ, CUDA_VISIBLE_DEVICES="", OMP_NUM_THREADS="1",
               OPENBLAS_NUM_THREADS="1", MKL_NUM_THREADS="1", NUMEXPR_NUM_THREADS="1")
    active = {}
    interrupted = False
    def stop(*args):
        nonlocal interrupted
        interrupted = True
    signal.signal(signal.SIGTERM, stop)
    while True:
        plan = json.loads(plan_path.read_text())
        end = (float("inf") if plan.get("reservation_managed_by_user")
               else datetime.fromisoformat(plan["reservation_end"]).timestamp())
        failures, finished, pending, waiting = [], [], [], []
        for key, (process, log) in list(active.items()):
            if process.poll() is not None:
                log.close()
                del active[key]
        for job in plan["jobs"]:
            key = f'{job["task"]}_{job["demos"]}_{job["seed"]}'
            path = directory / "states" / (key + ".json")
            state = json.loads(path.read_text()) if path.exists() else None
            if state and state["status"] == "completed":
                receipt = directory / "online" / (f'{job["task"]}_tapas_{job["demos"]}demos_seed{job["seed"]}.zip.json')
                if not receipt.exists():
                    publish_tapas_variant(plan, state, github_token)
                finished.append(key)
            elif state and state["status"] == "failed":
                failures.append(key)
            elif key not in active:
                if state:
                    failures.append(key + ": previous process missing; inspect before retry")
                elif time.time() + job["reservation_allowance_hours"] * 3600 < end and not interrupted:
                    pending.append(job)
                else:
                    waiting.append(key)
        while pending and len(active) < plan["training_workers"] and not interrupted:
            job = pending.pop(0)
            key = f'{job["task"]}_{job["demos"]}_{job["seed"]}'
            log = (directory / "logs" / (key + ".log")).open("a")
            process = subprocess.Popen([sys.executable, "-u", "-c", code,
                json.dumps([str(plan_path), job["task"], job["demos"], job["seed"]])],
                cwd=Path(plan["project"]) / "own_code/notebooks", env=env, stdout=log,
                stderr=subprocess.STDOUT, start_new_session=True)
            active[key] = process, log
        summary = dict(pid=os.getpid(), host=socket.gethostname(), updated=time.time(), planned=len(plan["jobs"]),
            state="training" if active else "awaiting_reservation_extension" if waiting else "finished",
            completed_online=finished, failed=failures, waiting=waiting,
            active={key:p.pid for key,(p,_) in active.items()}, reservation_end=plan["reservation_end"],
            reservation_managed_by_user=bool(plan.get("reservation_managed_by_user")))
        target = directory / "queue_state.json"
        target.with_suffix(".tmp").write_text(json.dumps(summary, indent=2))
        target.with_suffix(".tmp").replace(target)
        if time.time() >= end and active:
            for process, log in active.values():
                os.killpg(process.pid, signal.SIGTERM)
            raise RuntimeError("Reservation ended; own additional training workers were stopped")
        if not active and (not pending or interrupted):
            return summary
        time.sleep(15)
```

```python
def prioritize_tapas_training(plan_path):
    import json, os, time, signal, hashlib, subprocess, sys, fcntl
    from datetime import datetime
    from pathlib import Path
    plan_path = Path(plan_path)
    plan = json.loads(plan_path.read_text())
    directory = Path(plan["directory"])
    project = Path(plan["project"])
    audit_path = directory / "priority_state.json"
    lock = (directory / "priority.lock").open("a+")
    fcntl.flock(lock, fcntl.LOCK_EX | fcntl.LOCK_NB)
    queues = ["1f8b866815b9", "97e0beb3e556"]
    audit = dict(pid=os.getpid(), started=time.time(), state="draining_supplemental_evaluations",
                 reason="User priority: finish and publish new TAPAS models before supplemental evaluations",
                 training_workers=20, queues={})
    def save():
        audit["updated"] = time.time()
        temporary = audit_path.with_suffix(".tmp")
        temporary.write_text(json.dumps(audit, indent=2))
        temporary.replace(audit_path)
    for queue in queues:
        folder = project / "evaluation_logs" / queue
        state = json.loads((folder / "queue_state.json").read_text())
        pid = state["pid"]
        process = Path("/proc") / str(pid)
        assert process.exists() and process.stat().st_uid == os.getuid(), queue
        command = process.joinpath("cmdline").read_bytes()
        assert b"run_saved_queue" in command and queue.encode() in command, queue
        audit["queues"][queue] = dict(previous_pid=pid,
            protocol_sha256=hashlib.sha256((folder / "protocol.json").read_bytes()).hexdigest())
    save()
    for entry in audit["queues"].values():
        os.kill(entry["previous_pid"], signal.SIGTERM)
    while any(Path("/proc", str(entry["previous_pid"])).exists() for entry in audit["queues"].values()):
        save()
        time.sleep(15)
    for queue in queues:
        state = json.loads((project / "evaluation_logs" / queue / "queue_state.json").read_text())
        assert state["state"] in ("interrupted", "finished"), (queue, state)
    plan = json.loads(plan_path.read_text())
    for job in plan["jobs"]:
        job["original_reservation_allowance_hours"] = job["reservation_allowance_hours"]
        job["reservation_allowance_hours"] = 0
    plan["training_workers"] = 20
    plan["priority_authorization"] = "2026-10-04: start long jobs now; user will extend reservation this evening"
    plan["reservation_rule"] = "Existing reservation_end remains enforced until extension is confirmed"
    plan["jobs"].sort(key=lambda job: -job["original_reservation_allowance_hours"])
    temporary = plan_path.with_suffix(".priority.tmp")
    temporary.write_text(json.dumps(plan, indent=2))
    temporary.replace(plan_path)
    audit["state"] = "training_priority_active"
    save()
    while True:
        plan = json.loads(plan_path.read_text())
        state = json.loads((directory / "queue_state.json").read_text())
        if not state["active"]:
            break
        if not Path("/proc", str(state["pid"])).exists():
            raise RuntimeError("Training coordinator ended unexpectedly; inspect remaining workers")
        if time.time() >= datetime.fromisoformat(plan["reservation_end"]).timestamp():
            audit["state"] = "reservation_ended"
            save()
            return audit
        save()
        time.sleep(30)
    cells = json.loads((project / "own_code/notebooks/evaluate.ipynb").read_text())["cells"]
    function = "".join(next(c["source"] for c in cells if c.get("id") == "distributed-evaluation"))
    prefix = "import os,sys,json,time,hashlib,subprocess,signal\nfrom pathlib import Path\nfrom copy import deepcopy\nfrom concurrent.futures import ThreadPoolExecutor\nimport numpy as np\nimport pandas as pd\n"
    prefix += f"PROJECT_DIR={str(project)!r}\n"
    for queue in queues:
        folder = project / "evaluation_logs" / queue
        assert hashlib.sha256((folder / "protocol.json").read_bytes()).hexdigest() == audit["queues"][queue]["protocol_sha256"]
        state = json.loads((folder / "queue_state.json").read_text())
        if state["state"] == "finished":
            continue
        assert not Path("/proc", str(state["pid"])).exists()
        owner = json.loads((Path("/data2/ws1/schmidtn-study-evaluation/distribution/owners") / (queue + ".json")).read_text())
        if time.time() >= datetime.fromisoformat(owner["stop_dispatch_at"]).timestamp():
            audit["queues"][queue]["resume"] = "awaiting_reservation_extension"
            continue
        with (directory / (queue + "_resume.log")).open("a") as log:
            process = subprocess.Popen([sys.executable, "-u", "-c", prefix + function + f"\nrun_saved_queue({queue!r}, gpu_workers=3)"],
                cwd=project, stdout=log, stderr=subprocess.STDOUT, start_new_session=True)
        audit["queues"][queue]["resume_pid"] = process.pid
    audit["state"] = "training_priority_finished"
    audit["training_failed"] = json.loads((directory / "queue_state.json").read_text()).get("failed", [])
    save()
    return audit
```

```python
def backup_training_demos(project, output_directory, embedded=None, include_datasets=True, dataset_root=None):
    import os, json, io, tarfile, pickle, hashlib, socket, time
    from datetime import datetime, timezone
    from pathlib import Path
    project, output_directory = Path(project), Path(output_directory)
    output_directory.mkdir(parents=True, exist_ok=True)
    stamp = datetime.now(timezone.utc).strftime("%Y%m%d_%H%M%S")
    archive = output_directory / f"demos_{socket.gethostname()}_{stamp}.tar.gz"
    temporary = archive.with_suffix(".part")
    manifest = dict(host=socket.gethostname(), created_utc=stamp, files={}, datasets={}, embedded_demos={},
        note="Recorded SceneDatasets and original embedded demonstrations; no trajectories predicted by learned models. Embedded Demos are not full original SceneDataset directories.")
    def add_bytes(tar, name, content, source):
        item = tarfile.TarInfo(name)
        item.size, item.mtime = len(content), int(time.time())
        tar.addfile(item, io.BytesIO(content))
        manifest["files"][name] = dict(bytes=len(content), sha256=hashlib.sha256(content).hexdigest(), source=source)
    with tarfile.open(temporary, "w:gz", compresslevel=1) as tar:
        if include_datasets:
            data = Path(dataset_root) if dataset_root is not None else project / "own_code/artifacts/datasets/tapas_diffusion"
            names = ["bimanual_handover_item", "bimanual_put_bottle_in_fridge", "bimanual_take_tray_out_of_oven"]
            if dataset_root is not None or (data / "bimanual_dual_push_buttons").is_dir():
                names.insert(0, "bimanual_dual_push_buttons")
            for name in names:
                folder = data / name
                trajectories = sorted(p.name for p in (folder / "trajectories").iterdir() if p.is_dir())
                assert len(trajectories) == 100, name
                manifest["datasets"][name] = dict(trajectories=trajectories, source=str(folder.resolve()))
                for path in sorted(folder.rglob("*")):
                    if path.is_file():
                        before = path.stat()
                        content = path.read_bytes()
                        after = path.stat()
                        assert (before.st_size, before.st_mtime_ns) == (after.st_size, after.st_mtime_ns), path
                        add_bytes(tar, "datasets/" + path.relative_to(data).as_posix(), content, str(path))
            for path in sorted(data.glob("*.json")):
                add_bytes(tar, "provenance/scene_bundle/" + path.name, path.read_bytes(), str(path))
        for label, path in (embedded or {}).items():
            path = Path(path)
            with path.open("rb") as file:
                demos = pickle.load(file)._demos
            content = pickle.dumps(demos, protocol=4)
            add_bytes(tar, "embedded_demos/" + label + ".pkl", content, str(path))
            manifest["embedded_demos"][label] = dict(trajectories=demos.n_trajs, source_checkpoint_sha256=hashlib.sha256(path.read_bytes()).hexdigest())
        for path in sorted((project / "training_runs").rglob("run.json")):
            if path.relative_to(project / "training_runs").parts[0] in ("push_button", "hand_over", "put_bottle_in_fridge", "take_tray_out_of_oven"):
                add_bytes(tar, "provenance/" + path.relative_to(project).as_posix(), path.read_bytes(), str(path))
        for path in (project / "own_code/notebooks").glob("train_*.ipynb"):
            add_bytes(tar, "provenance/" + path.name, path.read_bytes(), str(path))
        content = json.dumps(manifest, indent=2).encode()
        item = tarfile.TarInfo("demo_backup_manifest.json")
        item.size = len(content)
        tar.addfile(item, io.BytesIO(content))
    temporary.replace(archive)
    checksum = hashlib.sha256(archive.read_bytes()).hexdigest()
    receipt = dict(archive=str(archive), sha256=checksum, bytes=archive.stat().st_size,
        files=len(manifest["files"]), datasets={k:len(v["trajectories"]) for k,v in manifest["datasets"].items()},
        embedded_demos=manifest["embedded_demos"])
    archive.with_suffix(".receipt.json").write_text(json.dumps(receipt, indent=2))
    return receipt
```

```python
def extend_tapas_five_demo_training(plan_path, training_workers=18, prepare_only=False):
    import json, hashlib, pickle, time, fcntl
    from copy import deepcopy
    from types import SimpleNamespace
    import numpy as np
    plan_path = Path(plan_path)
    plan = json.loads(plan_path.read_text())
    directory, project = Path(plan["directory"]), Path(plan["project"])
    lock = (directory / "five_demo_extension.lock").open("a+")
    fcntl.flock(lock, fcntl.LOCK_EX | fcntl.LOCK_NB)
    for name, checksum in plan["source_sha256"].items():
        assert hashlib.sha256((project / name).read_bytes()).hexdigest() == checksum, name
    cells = json.loads((project / "own_code/notebooks/train_TAPAS_bimanual.ipynb").read_text())["cells"][:11]
    assert hashlib.sha256(json.dumps([c["source"] for c in cells], sort_keys=True).encode()).hexdigest() == plan["training_cells_sha256"]
    def trajectory_digest(demos, index):
        digest = hashlib.sha256()
        def add(value):
            if isinstance(value, (tuple, list)):
                digest.update(str(len(value)).encode())
                for part in value:
                    add(part)
            else:
                array = value.detach().cpu().numpy() if hasattr(value, "detach") else np.asarray(value)
                digest.update(str((array.shape, array.dtype)).encode())
                digest.update(array.tobytes())
        for name in ("ee_poses_raw", "gripper_states_left", "gripper_states_right"):
            add(getattr(demos, name)[index])
        return digest.hexdigest()
    originals = {}
    for n in (25, 100):
        path = directory / "demo_sources" / f"{n}demos/latest.pkl"
        assert hashlib.sha256(path.read_bytes()).hexdigest() == plan["push_demo_checkpoints"][str(n)]
        with path.open("rb") as file:
            originals[n] = pickle.load(file)._demos
        assert originals[n].n_trajs == n
    fingerprints = {n: [trajectory_digest(data, i) for i in range(n)] for n, data in originals.items()}
    assert len(set(fingerprints[25])) == 25 and len(set(fingerprints[100])) == 100
    assert set(fingerprints[25]).issubset(fingerprints[100])
    subset = deepcopy(originals[25])
    for name, value in vars(subset).copy().items():
        if isinstance(value, tuple) and len(value) == 25:
            setattr(subset, name, value[:5])
        elif name.startswith("stacked_"):
            setattr(subset, name, value[:5])
    subset.n_trajs = 5
    subset.min_traj_len, subset.max_traj_len = min(subset.traj_lens), max(subset.traj_lens)
    subset.mean_traj_len = int(np.mean(subset.traj_lens))
    subset.subsample_to_common_length()
    selected = [trajectory_digest(subset, i) for i in range(5)]
    assert selected == fingerprints[25][:5]
    assert subset.frame_names == originals[25].frame_names and subset.meta_data == originals[25].meta_data
    assert all(value.shape[0] == 5 for name, value in vars(subset).items() if name.startswith("stacked_"))
    source = directory / "demo_sources/5demos/latest.pkl"
    source.parent.mkdir(parents=True, exist_ok=True)
    if source.exists():
        with source.open("rb") as file:
            previous = pickle.load(file)._demos
        assert previous.n_trajs == 5 and [trajectory_digest(previous, i) for i in range(5)] == selected
    else:
        temporary = source.with_suffix(".part")
        with temporary.open("xb") as file:
            pickle.dump(SimpleNamespace(_demos=subset), file, protocol=4)
        temporary.replace(source)
    checksum = hashlib.sha256(source.read_bytes()).hexdigest()
    if "5" in plan["push_demo_checkpoints"]:
        assert plan["push_demo_checkpoints"]["5"] == checksum
    audit = dict(created=time.time(), phase_alignment="gripper", training_cells_sha256=plan["training_cells_sha256"],
        original_jobs=24, added_jobs=12, tasks=list(plan["tasks"]), seeds=[1, 2, 3], diagnostics_reused=False,
        push_demo_source=str(source), push_demo_sha256=checksum, selected_25_indices=list(range(5)),
        selected_100_indices=[fingerprints[100].index(h) for h in selected], trajectory_sha256=selected,
        original_data_only=True, no_generated_trajectories=True, prepare_only=prepare_only)
    audit_path = directory / ("five_demo_preparation.json" if prepare_only else "five_demo_authorization_20261005.json")
    audit_path.with_suffix(".tmp").write_text(json.dumps(audit, indent=2))
    audit_path.with_suffix(".tmp").replace(audit_path)
    if not prepare_only:
        existing = {(j["task"], j["demos"], j["seed"]) for j in plan["jobs"]}
        additions = [dict(task=t, demos=5, seed=s, reservation_allowance_hours=0)
            for t in plan["tasks"] for s in (1, 2, 3) if (t, 5, s) not in existing]
        plan["jobs"] = additions + plan["jobs"]
        assert len(plan["jobs"]) == len({(j["task"], j["demos"], j["seed"]) for j in plan["jobs"]}) == 36
        plan["push_demo_checkpoints"]["5"] = checksum
        plan.update(training_workers=training_workers, reservation_managed_by_user=True, five_demo_authorization=str(audit_path))
        plan_path.with_suffix(".tmp").write_text(json.dumps(plan, indent=2))
        plan_path.with_suffix(".tmp").replace(plan_path)
    lock.close()
    return audit
```

```python
def train_tapas_position_variant(plan_path, task, demos, seed):
    import json, hashlib
    from pathlib import Path
    import numpy as np
    from tapas_gmm.policy.models.tpgmm import AutoTPGMM, FrameSelectionConfig, FittingStage
    rule = "balanced_position_precision_init_default_threshold_v1"
    plan = json.loads(Path(plan_path).read_text())
    state_path = Path(plan["directory"]) / "states" / f"{task}_{demos}_{seed}.json"
    if state_path.exists():
        assert json.loads(state_path.read_text()).get("training_method") == rule
    old_config, old_fit = FrameSelectionConfig.__init__, AutoTPGMM._fit_frames
    threshold = FrameSelectionConfig().rel_score_threshold
    def configure(self, *args, **kwargs):
        kwargs.update(fitting_actions=[FittingStage.INIT], rel_score_threshold=threshold)
        return old_config(self, *args, **kwargs)
    def fit_frames(self, *args, **kwargs):
        models, full = old_fit(self, *args, **kwargs)
        start = int(self.config.tpgmm.add_time_component)
        log_precision = np.array([[[
            -np.linalg.slogdet(g.sigma[start:start + 3, start:start + 3])[1]
            for g in arm.model.gaussians] for arm in frame] for frame in models])
        position = np.exp(log_precision - log_precision.max(axis=0, keepdims=True))
        position /= position.sum(axis=0, keepdims=True)
        self._frame_selection_rule = rule
        return models, full / full.sum(axis=0, keepdims=True) + position
    FrameSelectionConfig.__init__, AutoTPGMM._fit_frames = configure, fit_frames
    try:
        result = train_tapas_variant(plan_path, task, demos, seed)
    finally:
        FrameSelectionConfig.__init__, AutoTPGMM._fit_frames = old_config, old_fit
    cells = json.loads(Path(plan["project"], "own_code/notebooks/train_TAPAS_bimanual.ipynb").read_text())["cells"]
    source = "".join(next(c["source"] for c in cells if c.get("id") == "tapas-position-frame-variant"))
    result.update(training_method=rule, training_method_sha256=hashlib.sha256(source.encode()).hexdigest(),
                  frame_selection_threshold=threshold, diagnostic_only=plan.get("diagnostic_only", False))
    for path in (state_path, Path(result["checkpoint"]).with_name("run.json")):
        path.with_suffix(".tmp").write_text(json.dumps(result, indent=2))
        path.with_suffix(".tmp").replace(path)
    return result
```

```python
def train_tapas_temporal_variant(plan_path, task, demos, seed):
    """Official temporal-frame cohort; use the normal first eleven training cells."""
    import os, json, hashlib, pickle, fcntl, time
    from pathlib import Path
    plan = json.loads(Path(plan_path).read_text())
    key = f"{task}_{demos}_{seed}"
    assert plan["cohort"] == "tapas_anomaly_repair_20261007"
    assert plan["segmentation"] == "gripper" and plan["frame_selection_threshold"] == 0.75
    assert any(j["task"] == task and j["demos"] == demos and j["seed"] == seed for j in plan["jobs"])
    directory = Path(plan["directory"])
    guard = (directory / "locks" / (key + ".training.lock")).open("a+")
    fcntl.flock(guard, fcntl.LOCK_EX | fcntl.LOCK_NB)
    try:
        result = train_tapas_variant(plan_path, task, demos, seed)
        assert result["cohort"] == plan["cohort"]
        with Path(result["checkpoint"]).open("rb") as handle:
            model = pickle.load(handle)
        assert model.config.demos_segmentation.gripper_based
        assert not model.config.demos_segmentation.velocity_based
        assert model.config.frame_selection.rel_score_threshold == 0.75
        assert model._frame_selection_rule == "temporal_xyz_pose_precision_deduplicated_v1"
        assert model._demos.n_trajs == demos
        import numpy as np
        numerical_checks = []
        for index, segment in enumerate(model.segment_gmms):
            hmm = segment.model
            assert np.isfinite(hmm.mu).all(), 'Nonfinite learned means'
            assert np.isfinite(hmm.sigma).all(), 'Nonfinite learned covariance'
            eigenvalues = np.linalg.eigvalsh(hmm.sigma)
            assert np.min(eigenvalues) > 0, 'Nonpositive learned covariance'
            assert np.isfinite(hmm.Trans).all() and (hmm.Trans >= 0).all()
            assert np.allclose(hmm.Trans.sum(axis=1), 1), 'Transition rows are not normalized'
            assert np.isfinite(hmm.init_priors).all() and (hmm.init_priors >= 0).all()
            assert np.isclose(hmm.init_priors.sum(), 1), 'Initial priors are not normalized'
            numerical_checks.append(dict(segment=index, states=len(hmm.init_priors),
                minimum_covariance_eigenvalue=float(np.min(eigenvalues)),
                maximum_transition_row_error=float(np.max(np.abs(hmm.Trans.sum(axis=1)-1))),
                all_parameters_finite=True))
        segmentation_audit = getattr(model._demos, 'gripper_segmentation_audit', None)
        assert segmentation_audit and segmentation_audit['implementation'] == 'recorded_gripper_state_min_length_v1'
        assert segmentation_audit['demonstrations_preserved'] == demos
        result.update(training_method=model._frame_selection_rule, segmentation="gripper",
            numerical_implementation='log_domain_hmm_source_row_em_v1',
            gripper_segmentation_audit=segmentation_audit, numerical_checks=numerical_checks,
            replaces_configuration=plan['replaces_configuration'],
            gripper_based=True, velocity_based=False, frame_selection_threshold=0.75,
            diagnostic_only=False, official_benchmark_model=True, online=False,
            online_backup_state="pending_separate_publisher",
            verified_time=time.time(), host=plan["host"],
            dataset_provenance=plan["dataset_provenance"].get(task),
            selected_frames_audit=getattr(model, "frame_selection_audit", []))
        for path in (directory / "states" / (key + ".json"), Path(result["checkpoint"]).with_name("run.json")):
            temporary = path.with_suffix(".temporal.tmp")
            temporary.write_text(json.dumps(result, indent=2, allow_nan=False))
            temporary.replace(path)
        return result
    finally:
        guard.close()


def run_tapas_temporal_training(plan_path):
    """CPU queue with disjoint host assignments; stop admission, never kill active training."""
    import os, sys, json, time, hashlib, subprocess, socket, signal, fcntl, threading
    from pathlib import Path
    plan_path = Path(plan_path)
    plan = json.loads(plan_path.read_text())
    directory = Path(plan["directory"])
    assert socket.gethostname() == plan["host"]
    assigned = {(j["task"], j["demos"], j["seed"]) for j in plan["jobs"]}
    assert len(assigned) == 3 and len(plan["jobs"]) == 3
    assert all(plan["assignment"][j["task"]] == plan["host"] for j in plan["jobs"])
    assert {j["seed"] for j in plan["jobs"]} == {1, 2, 3}
    assert {j["demos"] for j in plan["jobs"]} == set(plan['evaluation_demos'])
    assert len(plan['evaluation_tasks']) == len(plan['evaluation_demos']) == 1
    for name in ("states", "logs", "locks", "tmp", "archives"):
        (directory / name).mkdir(exist_ok=True)
    guard = (directory / "training.lock").open("a+")
    fcntl.flock(guard, fcntl.LOCK_EX | fcntl.LOCK_NB)
    guard.seek(0); guard.truncate(); guard.write(str(os.getpid())); guard.flush()
    notebook = Path(plan["project"]) / "own_code/notebooks/train_TAPAS_bimanual.ipynb"
    cells = {c.get("id"): "".join(c.get("source", [])) for c in json.loads(notebook.read_text())["cells"]}
    assert hashlib.sha256(cells["tapas-temporal-cohort-training"].encode()).hexdigest() == plan["training_driver_sha256"]
    code = "import os,random,json,sys\nfrom pathlib import Path\n"
    code += "sys.path[:0] = " + repr([plan["project"] + "/" + p for p in ("TAPAS", "riepybdlib")]) + "\n"
    code += "nb=json.loads(Path(" + repr(str(notebook)) + ").read_text())\n"
    code += "cells={c.get('id'):''.join(c.get('source',[])) for c in nb['cells']}\n"
    code += "exec(cells['tapas-benchmark-training'])\nexec(cells['tapas-temporal-cohort-training'])\n"
    code += "train_tapas_temporal_variant(*json.loads(sys.argv[1]))\n"
    env = dict(os.environ, CUDA_VISIBLE_DEVICES="", OMP_NUM_THREADS="1", OPENBLAS_NUM_THREADS="1",
        MKL_NUM_THREADS="1", NUMEXPR_NUM_THREADS="1", TMPDIR=str(directory / "tmp"),
        TMP=str(directory / "tmp"), TEMP=str(directory / "tmp"), PYTHONDONTWRITEBYTECODE="1")
    stopping = threading.Event()
    old_handler = signal.signal(signal.SIGTERM, lambda *args: stopping.set())
    active, history, verified = {}, {}, {}
    identity = dict(pid=os.getpid(), uid=os.getuid(), host=socket.gethostname(),
        start_ticks=Path("/proc/self/stat").read_text().rsplit(")", 1)[1].split()[19])
    def save(summary):
        target = directory / "queue_state.json"
        temporary = target.with_suffix(".tmp")
        temporary.write_text(json.dumps(dict(summary, **identity, updated=time.time()), indent=2))
        temporary.replace(target)
    try:
        while True:
            completed, failed, pending = [], [], []
            for key, (process, log) in list(active.items()):
                if process.poll() is not None:
                    history[key].update(exit_code=process.returncode, finished=time.time())
                    log.close(); del active[key]
            for job in plan["jobs"]:
                key = f'{job["task"]}_{job["demos"]}_{job["seed"]}'
                path = directory / "states" / (key + ".json")
                state = json.loads(path.read_text()) if path.exists() else None
                if state and state.get("status") == "completed" and state.get("official_benchmark_model"):
                    assert state["cohort"] == plan["cohort"]
                    assert state.get("segmentation") == "gripper" and state.get("official_benchmark_model")
                    if key not in verified:
                        assert hashlib.sha256(Path(state["checkpoint"]).read_bytes()).hexdigest() == state["checkpoint_sha256"]
                        verified[key] = state["checkpoint_sha256"]
                    assert verified[key] == state["checkpoint_sha256"]
                    completed.append(key)
                elif key in active:
                    continue
                elif state:
                    failed.append(dict(job=key, state=state.get("status"), error=state.get("error"),
                        reason="Saved incomplete/failed attempt requires inspection; no blind retry"))
                elif key in history:
                    failed.append(dict(job=key, reason="Worker ended without a complete state", **history[key]))
                else:
                    pending.append(job)
            while pending and len(active) < plan["training_workers"] and not stopping.is_set():
                job = pending.pop(0)
                key = f'{job["task"]}_{job["demos"]}_{job["seed"]}'
                log = (directory / "logs" / (key + ".log")).open("a")
                process = subprocess.Popen([sys.executable, "-u", "-c", code,
                    json.dumps([str(plan_path), job["task"], job["demos"], job["seed"]])],
                    cwd=Path(plan["project"]) / "own_code/notebooks", env=env, stdout=log,
                    stderr=subprocess.STDOUT, stdin=subprocess.DEVNULL, start_new_session=True)
                parts = Path("/proc", str(process.pid), "stat").read_text().rsplit(")", 1)[1].split()
                active[key] = process, log
                history[key] = dict(pid=process.pid, uid=os.getuid(), start_ticks=parts[19], started=time.time())
            state_name = ("draining" if stopping.is_set() and active else
                "training" if active else "needs_attention" if failed else
                "interrupted" if stopping.is_set() and len(completed) < len(plan["jobs"]) else "finished")
            summary = dict(cohort=plan["cohort"], state=state_name, planned=len(plan["jobs"]),
                completed_models=completed, completed_online=[], failed=failed,
                active={key: process.pid for key, (process, _) in active.items()},
                child_start_ticks={key: history[key]["start_ticks"] for key in active},
                pending=len(pending), online_backup_state="pending_separate_publisher",
                segmentation="gripper", training_method="temporal_xyz_pose_precision_deduplicated_v1",
                reservation_managed_by_user=True, assigned_tasks=plan["evaluation_tasks"])
            save(summary)
            if not active and (not pending or stopping.is_set()):
                return summary
            stopping.wait(10) if not stopping.is_set() else time.sleep(2)
    except BaseException as error:
        save(dict(state="needs_attention", error=repr(error), planned=len(plan["jobs"]),
            active={key: process.pid for key, (process, _) in active.items()}))
        raise
    finally:
        signal.signal(signal.SIGTERM, old_handler)
        guard.close()
```
