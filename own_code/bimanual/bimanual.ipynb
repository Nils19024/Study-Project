```python
# Necessary modules and classes 
from rlbench.environment import Environment
from rlbench.action_modes.action_mode import BimanualMoveArmThenGripper
from rlbench.action_modes.arm_action_modes import BimanualEndEffectorPoseViaPlanning
from rlbench.action_modes.gripper_action_modes import BimanualDiscrete
from rlbench.bimanual_tasks.bimanual_dual_push_buttons import BimanualDualPushButtons
```

```python
# Environment creation with specified action mode and robot setup
env = Environment(
    action_mode=BimanualMoveArmThenGripper(
        BimanualEndEffectorPoseViaPlanning(
            absolute_mode=False, 
            frame="end effector",
        ),
        BimanualDiscrete()
    ),
    robot_setup="dual_panda",
    headless=True,
)

# Environment launch
env.launch()

# Add task to the environment
task = env.get_task(BimanualDualPushButtons)
#descriptions, obs = task.reset()

# Get one demonstration of the task
demos = task.get_demos(1, live_demos=True)

```

```python
observation1 = demos[0]._observations[0]
observation2 = demos[0]._observations[1]

```

```python
from tapas_gmm.env.rlbench import RLBenchEnvironment
```

```python
tapas_env = object.__new__(RLBenchEnvironment)
tapas_env.cameras = tuple()
tapas_env.task_env = task
```

```python
scene_observation = tapas_env.process_observation(observation1)
```

```python
action = tapas_env._get_action(observation1, observation2)
```

```python
tapas_env._delta_pos_scale = 0.01
tapas_env._delta_angle_scale = 0.04
obs, reward, done, info = tapas_env._step(
    action,
    postprocess=True,
    delay_gripper=False,
    scale_action=False,
)
```

```python
print("action shape:", action.shape)
print("reward:", reward)
print("done:", done)
```

```python
# Shutdown of the environment
env.shutdown()
```
